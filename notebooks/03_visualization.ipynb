# Medical Data Visualization

Create publication-ready medical visualizations including survival curves, distributions, and interactive plots.

```python
import sys
sys.path.insert(0, '..')

from src.visualization import MedicalVisualizer
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

visualizer = MedicalVisualizer(style='whitegrid', font_scale=1.1)
```

## 1. Create Sample Data

```python
# Generate sample medical data
np.random.seed(42)

n_samples = 200
data = {
    'age': np.random.normal(55, 15, n_samples),
    'treatment': np.random.choice(['Control', 'Treatment'], n_samples),
    'blood_pressure': np.random.normal(130, 15, n_samples),
    'cholesterol': np.random.normal(200, 40, n_samples),
    'bmi': np.random.normal(25, 4, n_samples)
}

df = pd.DataFrame(data)
print("Data shape:", df.shape)
print(df.head())
```

## 2. Distribution Plots

```python
# Age distribution
fig = visualizer.distribution_plot(df['age'], title='Age Distribution', bins=30)
plt.show()

# Cholesterol distribution
fig = visualizer.distribution_plot(df['cholesterol'], title='Cholesterol Distribution', bins=25)
plt.show()
```

## 3. Box and Violin Plots

```python
# Box plot by treatment group
fig = visualizer.box_plot(df, y='blood_pressure', x='treatment', 
                          title='Blood Pressure by Treatment Group')
plt.show()

# Violin plot by treatment group
fig = visualizer.violin_plot(df, y='cholesterol', x='treatment',
                             title='Cholesterol by Treatment Group')
plt.show()
```

## 4. Correlation Heatmap

```python
# Calculate correlation matrix
numeric_data = df[['age', 'blood_pressure', 'cholesterol', 'bmi']]
corr_matrix = numeric_data.corr()

# Create heatmap
fig = visualizer.heatmap(corr_matrix, title='Medical Variables Correlation Matrix')
plt.show()
```

## 5. Scatter Plots with Trend Lines

```python
# Age vs Cholesterol
fig = visualizer.scatter_with_trend(df['age'], df['cholesterol'], 
                                     title='Age vs Cholesterol Level')
plt.show()

# Age vs Blood Pressure
fig = visualizer.scatter_with_trend(df['age'], df['blood_pressure'],
                                     title='Age vs Blood Pressure')
plt.show()
```

## 6. Interactive Visualizations

```python
# Interactive scatter plot
fig = visualizer.interactive_scatter(df, x='age', y='cholesterol', 
                                     color='treatment',
                                     title='Age vs Cholesterol (Interactive)')
fig.show()
```

```python
# Interactive box plot
fig = visualizer.interactive_box(df, y='blood_pressure', x='treatment',
                                 title='Blood Pressure by Treatment (Interactive)')
fig.show()
```
