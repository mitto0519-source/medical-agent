# Automated Manuscript Generation

Generate structured medical manuscript sections automatically using the manuscript generator.

```python
import sys
sys.path.insert(0, '..')

from src.papergen import ManuscriptGenerator, CitationFormatter
from src.statistics import MedicalStatistics
import pandas as pd
import numpy as np
```

## 1. Initialize Manuscript Generator

```python
# Create manuscript generator
title = "Efficacy and Safety of Novel Treatment for Hypertension: A Randomized Controlled Trial"
authors = ["Dr. John Smith", "Dr. Jane Doe", "Dr. Michael Johnson"]
institution = "Medical Research Institute"

manuscript = ManuscriptGenerator(title, authors, institution)

print(f"Title: {manuscript.title}")
print(f"Authors: {', '.join(manuscript.authors)}")
print(f"Institution: {manuscript.institution}")
```

## 2. Generate Abstract

```python
background = "Hypertension affects millions globally and requires novel therapeutic approaches."
objective = "To evaluate the efficacy and safety of a novel antihypertensive medication."
methods = "200 patients were randomized to treatment (n=100) or control (n=100). Blood pressure was measured at baseline and 12 weeks."
results = "The treatment group showed significant reduction in systolic blood pressure (mean: -22.5 mmHg, p<0.001) compared to control (-3.2 mmHg)."
conclusion = "The novel treatment demonstrates superior efficacy in reducing blood pressure with acceptable safety profile."

abstract = manuscript.generate_abstract(background, objective, methods, results, conclusion)
print("\n" + abstract)
```

## 3. Generate Introduction

```python
problem = "current antihypertensive medications have limited efficacy in 30% of patients."
literature = "Recent studies highlight the role of endothelin receptor antagonism in blood pressure regulation."
gap = "the optimal combination therapy for resistant hypertension."
hypothesis = "combined therapy will achieve superior blood pressure control."

introduction = manuscript.generate_introduction(problem, literature, gap, hypothesis)
print("\n" + introduction)
```

## 4. Generate Methods

```python
design = "Randomized, double-blind, placebo-controlled trial conducted from 2023-2024."
population = "Adults aged 30-75 with stage 2 hypertension (SBP ≥160 mmHg). Exclusion: secondary hypertension, kidney disease."
collection = "Blood pressure measured using calibrated digital devices at baseline, 4, 8, and 12 weeks."
analysis = "Student's t-test for continuous variables, chi-square for categorical. Two-sided p<0.05 considered significant."

methods = manuscript.generate_methods(design, population, collection, analysis)
print("\n" + methods)
```

## 5. Generate Results

```python
# Create sample descriptive statistics
descriptive_stats = {
    'Age (mean ± SD)': '52.3 ± 11.2 years',
    'Gender (% female)': '48%',
    'Baseline SBP (mean ± SD)': '172.5 ± 15.3 mmHg',
    'Baseline DBP (mean ± SD)': '105.2 ± 10.1 mmHg',
    'Completion rate': '96% (192/200)'
}

findings = [
    'Treatment group achieved mean SBP reduction of 22.5 mmHg at 12 weeks (p<0.001)',
    'Control group achieved mean SBP reduction of 3.2 mmHg (p=0.23)',
    'Between-group difference was 19.3 mmHg (95% CI: 14.8-23.8, p<0.001)',
    'Treatment group patients had 68% response rate vs 12% in control group',
    'Adverse events were mild and similar between groups (p=0.67)'
]

results = manuscript.generate_results(descriptive_stats, findings)
print("\n" + results)
```

## 6. Generate Discussion

```python
interpretation = "Our findings demonstrate that the novel treatment achieves blood pressure reduction superior to standard therapy, with sustained effect throughout the 12-week period."
implications = "These results suggest novel treatment could benefit patients with inadequate response to conventional antihypertensives, potentially reducing cardiovascular risk."
limitations = "Single-center design; relatively short follow-up period; predominantly hypertensive population."
future = "Long-term studies (>12 months) and combination therapy trials are warranted."

discussion = manuscript.generate_discussion(interpretation, implications, limitations, future)
print("\n" + discussion)
```

## 7. Generate Complete Manuscript

```python
# Compile all sections
sections = {
    'abstract': "ABSTRACT\n" + abstract,
    'introduction': introduction,
    'methods': methods,
    'results': results,
    'discussion': discussion,
    'conclusion': manuscript.generate_conclusion(
        "This randomized trial demonstrates the efficacy of novel treatment in reducing blood pressure.",
        "These findings support further clinical development of this therapeutic approach."
    )
}

# Generate full manuscript
full_manuscript = manuscript.generate_full_manuscript(sections)
print(full_manuscript)
```

```python
# Example citations
author = "Smith J et al"
year = 2023
title = "Novel antihypertensive therapy in resistant hypertension"
journal = "Journal of Hypertension"
volume = 41
pages = "1234-1245"

formatter = CitationFormatter()

print("APA Style:")
print(formatter.format_apa(author, year, title, journal, volume, pages))

print("\nMLA Style:")
print(formatter.format_mla(author, title, journal, year, volume, pages))

print("\nChicago Style:")
print(formatter.format_chicago(author, title, journal, year, volume, pages))
```
