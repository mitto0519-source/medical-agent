# Medical Statistical Analysis

Comprehensive statistical tests for medical research including t-tests, chi-square, ANOVA, and non-parametric tests.

```python
import sys
sys.path.insert(0, '..')

from src.statistics import MedicalStatistics
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
plt.rcParams['figure.dpi'] = 100
```

## 1. Create Sample Medical Data

```python
# Generate sample data
np.random.seed(42)

n_patients = 150
data = {
    'patient_id': range(1, n_patients + 1),
    'age': np.random.normal(55, 15, n_patients),
    'treatment': np.random.choice(['Control', 'Treatment A', 'Treatment B'], n_patients),
    'baseline_score': np.random.normal(50, 15, n_patients),
    'followup_score': np.random.normal(55, 15, n_patients),
    'outcome': np.random.choice(['Improved', 'Stable', 'Declined'], n_patients)
}

df = pd.DataFrame(data)
print("Data shape:", df.shape)
print(df.head())
```

## 2. T-Test Analysis

```python
# Independent samples t-test
control = df[df['treatment'] == 'Control']['followup_score']
treatment_a = df[df['treatment'] == 'Treatment A']['followup_score']

med_stats = MedicalStatistics()
t_test_results = med_stats.t_test(control, treatment_a, paired=False)

print("Independent Samples T-Test Results:")
print(f"t-statistic: {t_test_results['t_statistic']:.4f}")
print(f"p-value: {t_test_results['p_value']:.4f}")
print(f"Cohen's d: {t_test_results['cohens_d']:.4f}")
print(f"Significant (p < 0.05): {t_test_results['significant']}")
```

## 3. ANOVA Test

```python
# One-way ANOVA for multiple groups
control_group = df[df['treatment'] == 'Control']['followup_score']
treatment_a_group = df[df['treatment'] == 'Treatment A']['followup_score']
treatment_b_group = df[df['treatment'] == 'Treatment B']['followup_score']

anova_results = med_stats.anova(control_group, treatment_a_group, treatment_b_group)

print("One-Way ANOVA Results:")
print(f"F-statistic: {anova_results['f_statistic']:.4f}")
print(f"p-value: {anova_results['p_value']:.4f}")
print(f"Significant (p < 0.05): {anova_results['significant']}")
```

## 4. Chi-Square Test

```python
# Create contingency table
contingency = pd.crosstab(df['treatment'], df['outcome'])
print("Contingency Table:")
print(contingency)

chi_square_results = med_stats.chi_square(contingency)

print("\nChi-Square Test Results:")
print(f"Chi-square statistic: {chi_square_results['chi2_statistic']:.4f}")
print(f"p-value: {chi_square_results['p_value']:.4f}")
print(f"Significant (p < 0.05): {chi_square_results['significant']}")
```

## 5. Non-Parametric Tests

```python
# Mann-Whitney U test (non-parametric alternative to t-test)
mw_results = med_stats.mann_whitney_u(control, treatment_a)

print("Mann-Whitney U Test Results:")
print(f"U-statistic: {mw_results['u_statistic']:.4f}")
print(f"p-value: {mw_results['p_value']:.4f}")
print(f"Significant (p < 0.05): {mw_results['significant']}")

# Kruskal-Wallis test (non-parametric alternative to ANOVA)
kw_results = med_stats.kruskal_wallis(control_group, treatment_a_group, treatment_b_group)

print("\nKruskal-Wallis Test Results:")
print(f"H-statistic: {kw_results['h_statistic']:.4f}")
print(f"p-value: {kw_results['p_value']:.4f}")
print(f"Significant (p < 0.05): {kw_results['significant']}")
```

## 6. Correlation Analysis

```python
# Pearson correlation
pearson_corr = med_stats.correlation(df['age'], df['followup_score'], method='pearson')

print("Pearson Correlation Results:")
print(f"Correlation coefficient: {pearson_corr['correlation']:.4f}")
print(f"p-value: {pearson_corr['p_value']:.4f}")
print(f"Significant (p < 0.05): {pearson_corr['significant']}")

# Spearman correlation (non-parametric)
spearman_corr = med_stats.correlation(df['age'], df['followup_score'], method='spearman')

print("\nSpearman Correlation Results:")
print(f"Correlation coefficient: {spearman_corr['correlation']:.4f}")
print(f"p-value: {spearman_corr['p_value']:.4f}")
```

## 7. Visualization of Results

```python
# Visualize group comparisons
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Box plot for treatment groups
df.boxplot(column='followup_score', by='treatment', ax=axes[0])
axes[0].set_title('Follow-up Scores by Treatment Group')
axes[0].set_xlabel('Treatment Group')
axes[0].set_ylabel('Follow-up Score')

# Bar plot for outcomes
outcome_counts = df['outcome'].value_counts()
axes[1].bar(outcome_counts.index, outcome_counts.values, color=['green', 'gray', 'red'])
axes[1].set_title('Patient Outcomes Distribution')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```
